### Examples of the dataset

Some examples of the dataset are shown below. Each image is a 28x28 grayscale image representing a handwritten digit.

```python
def show_examples(x_data, y_data, num_classes=10, num_photos=5, title=None):
    """
    Displays a grid of images from the dataset.

    Parameters:
    - x_data: The input data (images) to display.
    - y_data: The corresponding labels for the images.
    - num_classes: Number of classes to display (default is 10 for MNIST).
    - num_photos: Number of photos per class to display (default is 5).
    - title: Title for the entire figure (default is None).
    """

    fig, axes = plt.subplots(num_classes, num_photos, figsize=(num_photos * 2, num_classes * 2))

    if title:
        fig.suptitle(title, fontsize=16, fontweight='bold')

    for digit in range(num_classes):
        # Get indices of all images belonging to the current digit
        digit_indices = np.where(y_data == digit)[0]

        for photo_idx in range(num_photos):
            # Select the current index for the digit
            current_idx = digit_indices[photo_idx]

            # Select the appropriate axis for the current digit and photo index
            ax = axes[digit, photo_idx]

            ax.imshow(x_data[current_idx], cmap=CMAP)
            ax.axis('off')  # No need to see the axis for each image

            if photo_idx == 0:
                ax.set_title(f'Digit {digit}', loc='left', fontsize=12, fontweight='bold')

    plt.tight_layout()
    plt.show()


show_examples(x_train_pre, y_train, num_classes=10, num_photos=7, title="Examples of the MNIST Training Set (Nothing applied)")
```

### Image Redimensioning

Images are redimensioned, in case of using a smaller size, to reduce the number of parameters in the model and speed up training.

```python
def resize_images(images, new_size=(28, 28)):
    """
    Resize a batch of images to a new size using OpenCV.

    Parameters:
    - images: A numpy array of shape (num_images, height, width) containing the images to be resized.
    - new_size: A tuple (new_height, new_width) specifying the desired size of the output images.

    Returns:
    - A numpy array of shape (num_images, new_height, new_width) containing the resized images.
    """


    num_imgs = images.shape[0]
    new_width, new_height = new_size


    resized = np.zeros((num_imgs, new_height, new_width), dtype=np.float32)
    for i in range(num_imgs):
        resized[i] = cv2.resize(images[i].astype(np.float32), (new_width, new_height))
    return resized

NEW_WIDTH, NEW_HEIGHT = 28, 28
x_train_resized = resize_images(x_train_pre,    (NEW_WIDTH, NEW_HEIGHT))
x_val_resized = resize_images(x_val_pre,        (NEW_WIDTH, NEW_HEIGHT))


show_examples(x_train_resized, y_train, num_classes=10, num_photos=7, title=f"Examples of the MNIST Training Set (Resized to {NEW_WIDTH}x{NEW_HEIGHT})")

```

## Creating the Variational Autoencoder (VAE) Models

### Create the VAE

The model created in `vae.py` is used to create the Variational Autoencoder (VAE) model.
If possible, the model is moved to the GPU for faster training.

```python
LATENT_DIM = 10

gaussian_encoder = GaussianEncoder(input_dim=NEW_WIDTH*NEW_HEIGHT, hidden_dim=512, latent_dim=LATENT_DIM)
gaussian_decoder    = GaussianDecoder   (latent_dim=LATENT_DIM, hidden_dim=512, output_dim=NEW_WIDTH*NEW_HEIGHT)
bernoulli_decoder  = BernoulliDecoder  (latent_dim=LATENT_DIM, hidden_dim=512, output_dim=NEW_WIDTH*NEW_HEIGHT)

gaussian_vae    = VAE(gaussian_encoder, gaussian_decoder).to(device)
bernoulli_vae  = VAE(gaussian_encoder, bernoulli_decoder).to(device)
```

### Defining loss, optimizer and LR scheduler

The loss function is defined (see its comments for details).

The optimizer is defined using the Adam optimizer.

A scheduler is defined to reduce the learning rate when the validation loss plateaus. The scheduler is set to monitor the validation loss and reduce the learning rate by a factor of 0.5 if the validation loss does not improve for 5 consecutive epochs. The minimum learning rate is set to 0.0001.

```python
def loss_gaussian_vae(x, mu_latent, log_var_latent, mu_reconstruction, std_reconstruction=0.5):
    """
    Calculates the loss function for a Gaussian VAE.

    Parameters:
    - x: The original input data. Dimensions: (batch_size, input_dim)
    - mu_latent: The mean calculated by the encoder (mu_theta(x)). Dimensions: (batch_size, latent_dim)
    - log_var_latent: The logarithm of the variance from the encoder. Dimensions: (batch_size, latent_dim)
    - mu_reconstruction: The mean of the reconstruction distribution (p_theta(x|z)). Dimensions: (batch_size, input_dim)
    - std_reconstruction: The standard deviation of the reconstruction distribution.

    Returns:
    - total_loss: The total loss to minimize, which is the sum of the reconstruction loss and the KL divergence loss.
    - recon_loss: The reconstruction loss term.
    - kl_loss: The KL divergence loss term.
    """
    x_flattened = x.view(x.size(0), -1)  # Flatten x to match dimensions with mu_reconstruction

    # 1. Reconstruction Term: (1 / (2 * sigma^2)) * ||x - mu_phi(z)||^2
    recon_loss = F.mse_loss(mu_reconstruction, x_flattened, reduction='sum')
    recon_loss = recon_loss / (2.0 * std_reconstruction ** 2)

    # 2. KL Divergence Term (Closed-form solution for Gaussians)
    kl_loss = -0.5 * torch.sum(1 + log_var_latent - mu_latent.pow(2) - log_var_latent.exp())

    # Total loss to minimize
    total_loss = recon_loss + kl_loss

    return total_loss, recon_loss, kl_loss


def loss_bernoulli_vae(x, mu_latent, log_var_latent, p_reconstruction):
    """
    Calculates the loss function for a Bernoulli VAE.

    Parameters:
    - x: The original input data. Dimensions: (batch_size, input_dim)
    - mu_latent: The mean calculated by the encoder (mu_theta(x)). Dimensions: (batch_size, latent_dim)
    - log_var_latent: The logarithm of the variance from the encoder. Dimensions: (batch_size, latent_dim)
    - p_reconstruction: The probability of the reconstruction distribution (p_theta(x|z)). Dimensions: (batch_size, input_dim)

    Returns:
    - total_loss: The total loss to minimize, which is the sum of the reconstruction loss and the KL divergence loss.
    - recon_loss: The reconstruction loss term.
    - kl_loss: The KL divergence loss term.
    """
    x_flattened = x.view(x.size(0), -1)  # Flatten x to match dimensions with p_reconstruction

    # 1. Reconstruction Term: Binary Cross-Entropy Loss
    recon_loss = F.binary_cross_entropy(p_reconstruction, x_flattened, reduction='sum')

    # 2. KL Divergence Term (Closed-form solution for Gaussians)
    kl_loss = -0.5 * torch.sum(1 + log_var_latent - mu_latent.pow(2) - log_var_latent.exp())

    # Total loss to minimize
    total_loss = recon_loss + kl_loss

    return total_loss, recon_loss, kl_loss
```

```python
optimizer_gaussian = torch.optim.Adam(gaussian_vae.parameters(), lr=0.002)
scheduler_gaussian = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer_gaussian, mode='min', factor=0.5, patience=5, threshold=0.001, min_lr=0.0001
)

optimizer_bernoulli = torch.optim.Adam(bernoulli_vae.parameters(), lr=0.002)
scheduler_bernoulli = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer_bernoulli, mode='min', factor=0.5, patience=5, threshold=0.001, min_lr=0.0001
)
```

### Training the model

```python
def train_model(model, optimizer, scheduler, loss_function, trainloader, testloader, num_epochs=20, model_name="VAE"):
    """
    Trains the given VAE model.

    Parameters:
    - model: The VAE model to train.
    - optimizer: The optimizer for updating model weights.
    - scheduler: The learning rate scheduler.
    - loss_function: The loss function to use (either Gaussian or Bernoulli).
    - trainloader: DataLoader for the training dataset.
    - testloader: DataLoader for the validation/test dataset.
    - num_epochs: Number of epochs to train the model.
    - model_name: Name of the model (for logging purposes).

    Returns:
    - history: A dictionary containing training and validation losses over time. It contains:
        - 'train_recon_loss': List of average reconstruction losses for each training batch.
        - 'train_kl_loss': List of average KL divergence losses for each training batch.
        - 'total_steps_counter': List of step numbers corresponding to each training batch.
        - 'periodic_val_recon_loss': List of average reconstruction losses for validation batches (every PERIODIC_VALIDATION_INTERVAL steps).
        - 'periodic_val_kl_loss': List of average KL divergence losses for validation batches (every PERIODIC_VALIDATION_INTERVAL steps).
        - 'periodic_val_steps': List of step numbers corresponding to each validation batch.
    """


    print(f"Training model: {model_name} for {num_epochs} epochs on device: {device}")

    PERIODIC_VALIDATION_INTERVAL = 100  # Perform validation every 100 weight updates

    history = {
        'train_recon_loss': [], 'train_kl_loss': [],    # In each batch, the average reconstruction loss and KL divergence loss for that batch will be saved.
        'total_steps_counter': [],                      # In each batch, the step number will be saved, which will be used to plot the loss curves.
        'periodic_val_recon_loss': [], 'periodic_val_kl_loss': [],  # Every PERIODIC_VALIDATION_INTERVAL weight updates, the average reconstruction loss and KL divergence loss on a validation batch will be saved.
        'periodic_val_steps': []        # Every PERIODIC_VALIDATION_INTERVAL weight updates, the step number will be saved.
    }

    current_step_number = 0  # Tracks the number of weight updates performed so far

    for epoch in range(1, num_epochs + 1):
        model.train()
        running_epoch_loss = 0.0

        pbar = tqdm(trainloader, desc=f"Epoch {epoch:02d}/{num_epochs:02d}")

        for batch_idx, (data, _) in enumerate(pbar):
            data = data.to(device)
            batch_size = data.shape[0]
            optimizer.zero_grad()

            # Forward pass
            recon_batch, mu, log_var = model(data)

            # Calculate loss
            loss, recon_loss, kl_loss = loss_function(data, mu, log_var, recon_batch)

            # Backward pass (Backpropagation)
            loss.backward()

            # Update weights
            optimizer.step()

            # Save metrics for this batch
            running_epoch_loss += loss.item()
            recon_loss_normalized = recon_loss.item() / batch_size
            kl_loss_normalized = kl_loss.item() / batch_size
            history['train_recon_loss'].append(recon_loss_normalized)
            history['train_kl_loss'].append(kl_loss_normalized)

            current_step_number += 1
            history['total_steps_counter'].append(current_step_number)

            # --- Fast Validation every PERIODIC_VALIDATION_INTERVAL weight updates ---
            if current_step_number % PERIODIC_VALIDATION_INTERVAL == 0:
                model.eval()
                with torch.no_grad():
                    # Fast Validation => Only a single batch from the testloader is used to save time
                    val_data, _ = next(iter(testloader))
                    val_batch_size = val_data.shape[0]
                    val_data = val_data.to(device)

                    v_recon, v_mu, v_log_var = model(val_data)
                    _, v_recon_loss, v_kl_loss = loss_function(val_data, v_mu, v_log_var, v_recon)

                    # Save average loss for this validation batch
                    history['periodic_val_recon_loss'].append(v_recon_loss.item() / val_batch_size)
                    history['periodic_val_kl_loss'].append(v_kl_loss.item() / val_batch_size)
                    history['periodic_val_steps'].append(current_step_number)

                model.train()  # Switch back to training mode


            pbar.set_postfix({
                'Loss': f"{loss.item() / batch_size:.4f}",
                'Recon': f"{recon_loss_normalized:.4f}",
                'KL': f"{kl_loss_normalized:.4f}"
            })

        # --- End of Epoch: Full Evaluation on the Test Set ---
        model.eval()
        total_test_loss = 0.0

        with torch.no_grad():
            for test_data, _ in testloader:
                test_data = test_data.to(device)
                t_recon, t_mu, t_log_var = model(test_data)
                t_loss, _, _ = loss_function(test_data, t_mu, t_log_var, t_recon)
                total_test_loss += t_loss.item()

        epoch_train_avg = running_epoch_loss / len(trainloader.dataset)
        epoch_test_avg = total_test_loss / len(testloader.dataset)

        # Update learning rate based on average training loss
        scheduler.step(epoch_train_avg)

    # Save the trained model to disk
    DIR_TRAINED_MODELS = "trained_models"
    if not os.path.exists(DIR_TRAINED_MODELS):
        os.makedirs(DIR_TRAINED_MODELS)
    filename = os.path.join(DIR_TRAINED_MODELS, f"{model_name}_model.pth")
    torch.save(model.state_dict(), filename)

    print(f"Model {model_name} trained and saved to {filename}.")
    return history
```

```python
gaussian_name = "Gaussian VAE"
bernoulli_name = "Bernoulli VAE"

TRAINING = True
# Check for the trained models
if not os.path.exists("trained_models"):
    os.makedirs("trained_models")
    TRAINING = True

if TRAINING:
	set_seed(SEED)
	gaussian_vae_history    = train_model(gaussian_vae,     optimizer_gaussian,     scheduler_gaussian,     loss_gaussian_vae,  gaussian_trainloader,   gaussian_testloader,    num_epochs=20, model_name=gaussian_name)
	print("\n\n")
	set_seed(SEED)
	bernoulli_vae_history  = train_model(bernoulli_vae,   optimizer_bernoulli,   scheduler_bernoulli,   loss_bernoulli_vae, bernoulli_trainloader, bernoulli_testloader,  num_epochs=20, model_name=bernoulli_name)
else:
	# Load the trained models from disk
	gaussian_vae.load_state_dict(torch.load(os.path.join("trained_models", f"{gaussian_name}_model.pth"), map_location=device))
	bernoulli_vae.load_state_dict(torch.load(os.path.join("trained_models", f"{bernoulli_name}_model.pth"), map_location=device))
	gaussian_vae_history = None
	bernoulli_vae_history = None
	print(f"Models {gaussian_name} and {bernoulli_name} loaded from disk.")
```

### Evaluating the model

In order to evaluate the model, we will first generate some plots.
- The two loss components (reconstruction loss vs KL divergence) are plotted for both the training and validation sets.
- The total loss of the training and validation sets is plotted.

```python
gaussian_vae.eval()
bernoulli_vae.eval()

if not os.path.exists("results"):
    os.makedirs("results")


def plot_loss_components(history, model_name="VAE"):
    """
    Plots the training and validation loss components (Reconstruction Loss and KL Divergence) for a given VAE model.

    Parameters:
    - history: The corresponding history dictionary returned by the train_model function.
    - model_name: Name of the model (for title purposes).
    """

    plot_path = os.path.join("results", f"LossComponents_Comparison_{model_name.replace(' ', '')}.png")

    if TRAINING:
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5), dpi=100)
        fig.suptitle(f'{model_name} Training vs. Validation Loss Components Comparison', fontsize=14, fontweight='bold', y=1.02)

        # ---------------------------------------------------------------------
        # LEFT PLOT: Reconstruction Loss (Train vs. Validation)
        # ---------------------------------------------------------------------
        # Continuous line for frequent training steps
        ax1.plot(history['total_steps_counter'], history['train_recon_loss'],
                color='blue', alpha=0.6, label='Train Recon Loss')

        # Dashed line with markers for periodic validation steps
        ax1.plot(history['periodic_val_steps'], history['periodic_val_recon_loss'],
                color='red', linestyle='--', marker='o', markersize=4, label='Val Recon Loss')

        ax1.set_title('Reconstruction Error', fontsize=12, pad=10)
        ax1.set_xlabel('Training Steps', fontsize=10)
        ax1.set_ylabel('Loss Value', fontsize=10)
        ax1.grid(True, linestyle='--', alpha=0.5)
        ax1.legend(loc='upper right')

        # ---------------------------------------------------------------------
        # RIGHT PLOT: KL Divergence Loss (Train vs. Validation)
        # ---------------------------------------------------------------------
        # Continuous line for frequent training steps
        ax2.plot(history['total_steps_counter'], history['train_kl_loss'],
                color='green', alpha=0.6, label='Train KL Loss')

        # Dashed line with markers for periodic validation steps
        ax2.plot(history['periodic_val_steps'], history['periodic_val_kl_loss'],
                color='purple', linestyle='--', marker='o', markersize=4, label='Val KL Loss')

        ax2.set_title('KL Divergence (Latent Space Regularization)', fontsize=12, pad=10)
        ax2.set_xlabel('Training Steps', fontsize=10)
        ax2.set_ylabel('Loss Value', fontsize=10)
        ax2.grid(True, linestyle='--', alpha=0.5)
        ax2.legend(loc='upper right')

        # Adjust layout to prevent overlapping text
        plt.tight_layout()

        # Save the plot to disk
        plt.savefig(plot_path)
        plt.close(fig)


    # Load the saved plot from disk and display it
    img = cv2.imread(plot_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(15, 5), dpi=100)
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.show()


plot_loss_components(gaussian_vae_history,      model_name=gaussian_name)
plot_loss_components(bernoulli_vae_history,     model_name=bernoulli_name)
```

```python
gaussian_vae.eval()
bernoulli_vae.eval()

def plot_total_loss(history, model_name="VAE"):
    """
    Plots the total loss (ELBO) for both training and validation phases of a given VAE model.

    Parameters:
    - history: The corresponding history dictionary returned by the train_model function.
    - model_name: Name of the model (for title purposes).
    """

    plot_path = os.path.join("results", f"Total_Loss_Evolution_{model_name.replace(' ', '')}.png")

    if TRAINING:

        train_total_loss =  [- recon - kl for recon, kl in zip(history['train_recon_loss'], history['train_kl_loss'])]
        val_total_loss =    [- recon - kl for recon, kl in zip(history['periodic_val_recon_loss'], history['periodic_val_kl_loss'])]

        # Create the figure
        plt.figure(figsize=(8, 5), dpi=100)

        # Continuous line for frequent training steps
        plt.plot(history['total_steps_counter'], train_total_loss,
                color='green', alpha=0.5, label='Train Total Loss')

        # Disconnected horizontal dashes for periodic validation steps
        plt.plot(history['periodic_val_steps'], val_total_loss,
                color='red', linestyle='--', marker='o', markersize=4, label='Val Total Loss')

        # Aesthetics and labels
        plt.title(f'{model_name} Total Loss Evolution (ELBO): Train vs. Validation', fontsize=12, fontweight='bold', pad=12)
        plt.xlabel('Training Steps', fontsize=10)
        plt.ylabel('Total Loss Value', fontsize=10)
        plt.grid(True, linestyle='--', alpha=0.5)
        plt.legend(loc='lower right')

        plt.tight_layout()
        plt.savefig(plot_path)
        plt.close()

    img = cv2.imread(plot_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(8, 5), dpi=100)
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.show()

plot_total_loss(gaussian_vae_history, model_name=gaussian_name)
plot_total_loss(bernoulli_vae_history, model_name=bernoulli_name)
```

### Visualizating the image reconstruction

```python
def compare_vae_outputs(gaussian_loader, bernoulli_loader, gaussian_model, bernoulli_model, num_images=5, subtitle=""):
    """
    Plots a grid comparing normalized and binarized images
    against reconstructions from both Gaussian and Bernoulli VAEs,
    using separate dataloaders for each preprocessing pipeline.

    Parameters:
    - gaussian_loader: DataLoader for the normalized images (input to Gaussian VAE).
    - bernoulli_loader: DataLoader for the binarized images (input to Bernoulli VAE).
    - gaussian_model: The trained Gaussian VAE model.
    - bernoulli_model: The trained Bernoulli VAE model.
    - num_images: Number of images to visualize in the grid (default is 5).
    - subtitle: Optional subtitle for the entire figure (default is an empty string).
    """
    gaussian_model.eval()
    bernoulli_model.eval()

    # Use a fixed seed to ensure both iterators fetch the exact same batch of samples
    set_seed(SEED)
    gaussian_images, _  = next(iter(gaussian_loader))
    set_seed(SEED)
    bernoulli_images, _ = next(iter(bernoulli_loader))

    gaussian_images = gaussian_images.to(device)
    bernoulli_images = bernoulli_images.to(device)

    with torch.no_grad():
        gaussian_recon, _, _ = gaussian_model(gaussian_images)
        bernoulli_recon, _, _ = bernoulli_model(bernoulli_images)

    headers = ['1. Normalized (Gaussian In)', '2. Binarized (Bernoulli In)', '3. Bernoulli Output', '4. Gaussian Output']


    fig, axes = plt.subplots(num_images, len(headers), figsize=(12, num_images * 2.5), dpi=100)
    fig.suptitle(f'Image Preprocessing & VAE Reconstruction Comparison Grid\n{subtitle}',
                 fontsize=14, fontweight='bold', y=0.99)


    for row in range(num_images):
        img_norm = gaussian_images[row].cpu().numpy()
        img_bin = bernoulli_images[row].cpu().numpy()

        # --- Extract Model Outputs (Reshaped back to 2D matrices) ---
        img_bernoulli_out = bernoulli_recon[row].view(NEW_WIDTH, NEW_HEIGHT).cpu().numpy()
        img_gaussian_out = gaussian_recon[row].view(NEW_WIDTH, NEW_HEIGHT).cpu().numpy()

        # Consolidate the 4 states for the current row
        image_states = [img_norm, img_bin, img_bernoulli_out, img_gaussian_out]

        for col in range(len(headers)):
            ax = axes[row, col] if num_images > 1 else axes[col]

            # Use strict vmin/vmax to respect pixel value scales
            ax.imshow(image_states[col], cmap=CMAP, vmin=0, vmax=1)
            ax.axis('off')

            # Set headers only on the top row
            if row == 0:
                ax.set_title(headers[col], fontsize=10, fontweight='bold', pad=10)

    plt.tight_layout()

    plot_path = os.path.join("results", f"Reconstruction_{subtitle.replace(' ', '_')}.png")
    plt.savefig(plot_path)
    plt.show()

NUM_IMAGES_TO_COMPARE = 10
compare_vae_outputs(gaussian_trainloader, bernoulli_trainloader, gaussian_vae, bernoulli_vae, num_images=NUM_IMAGES_TO_COMPARE, subtitle=f"Training Set")
compare_vae_outputs(gaussian_testloader, bernoulli_testloader, gaussian_vae, bernoulli_vae, num_images=NUM_IMAGES_TO_COMPARE, subtitle=f"Testing Set")
```

### Data Interpolation in the latent space

Given the structure of the latent space learned by the VAE, we can perform interpolation between two points in the latent space. This allows us to generate a smooth transition between two different images.

```python
def interpolate_latent_space(vae_model, data_loader, idx_A=0, idx_B=1, num_steps=10, model_name="VAE"):
    """
    Generates a visual transition between two images in the latent space of a Gaussian VAE.

    Parameters:
    - vae_model: The trained VAE model.
    - data_loader: DataLoader for the dataset to sample images from.
    - idx_A: Index of the first image in the batch to interpolate from (default is 0).
    - idx_B: Index of the second image in the batch to interpolate to (default is 1).
    - num_steps: Number of interpolation steps between the two images (default is 10).
    - model_name: Name of the VAE model (default is "VAE").
    """

    vae_model.eval()

    images, labels = next(iter(data_loader))

    img_A = images[idx_A].to(device)
    img_B = images[idx_B].to(device)

    label_A = labels[idx_A].item()
    label_B = labels[idx_B].item()

    with torch.no_grad():
        # Flatten the images if the encoder expects a vector of size 784
        x_A = img_A.view(1, -1)
        x_B = img_B.view(1, -1)

        mu_A, _ = vae_model.encoder(x_A)
        mu_B, _ = vae_model.encoder(x_B)

    # Define the interpolation steps
    steps = np.linspace(0, 1, num_steps)

    fig, axes = plt.subplots(1, num_steps, figsize=(15, 2.5))

    for i, t in enumerate(steps):
        # Linear interpolation formula: (1 - t) * A + t * B
        interpolated_vector = (1.0 - t) * mu_A + t * mu_B

        with torch.no_grad():
            # Pass the interpolated coordinate directly to the Decoder
            decoded_image = vae_model.decoder(interpolated_vector)

            # Reshape the output to 2D format (28x28 pixels) for visualization
            image_numpy = decoded_image.view(NEW_HEIGHT, NEW_WIDTH).cpu().numpy()

        # Plot the resulting image in its respective subplot
        axes[i].imshow(image_numpy, cmap=CMAP, vmin=0, vmax=1)
        axes[i].axis('off')
        axes[i].set_title(f"t={t:.1f}", fontsize=9)


    title = f"Latent Space Interpolation ({model_name})"
    plt.suptitle(f"{title}\n Transition from digit {label_A} to digit {label_B}", fontsize=12, fontweight='bold', y=1.05)
    plt.tight_layout()

    plot_path = os.path.join("results", f"LatentSpace_Interpolation_{model_name.replace(' ', '')}.png")
    plt.savefig(plot_path)
    plt.show()

interpolate_latent_space(gaussian_vae, gaussian_trainloader, idx_A=0, idx_B=1, num_steps=10, model_name=gaussian_name)
interpolate_latent_space(bernoulli_vae, bernoulli_trainloader, idx_A=0, idx_B=1, num_steps=10, model_name=bernoulli_name)
```

## Exact Likelihood Estimation

The main aim of every DGM is to maximize the log-likelihood of the data, $\log p(x)$. For a VAE, the exact formula for the marginal likelihood is given by:
$$
\log p(x) = \log \int p(x|z)p(z) dz
$$

However, this integral is often intractable, as it requires summing over all possible latent variable configurations. Therefore, we will explore four different estimators for $\log p(x)$.

### 1. Sampling $z\in \mathbb{R}^d$.

In this case, $z$ is sampled uniformly from a bounded hypercube $\mathcal{Z} = [a, b]^d \subset \mathbb{R}^d$, where $d$ is the dimension of the latent space (`LATENT_DIM`). The distribution used for sampling $z$ is a uniform distribution over the hypercube:
$$p(z) = \frac{1}{\text{Volume}} \quad \text{for } z \in [a, b]^d$$
where $\text{Volume} = (b - a)^d$ is the volume of the hypercube. The Monte Carlo estimator for the likelihood is then given by:
$$
p(x) \approx \frac{\text{Volume}}{N} \sum_{i=1}^{N} p(x|z_i) \cdot p(z_i)
$$

Since probabilities are in the interval $[0, 1]$, we usually work with the log-likelihood instead, as it is in $]-\infty, 0]$, and is numerically more stable. The log-likelihood can be computed as:
$$
\log p(x) \approx \log \text{Volume} - \log N + \log \left( \sum_{i=1}^{N} p(x|z_i) \cdot p(z_i) \right)
$$

Since the probabilities $p(x|z_i)$ and $p(z_i)$ are also very small, they are obtained as log-probabilities. Therefore, the log-likelihood can be computed as:
$$
\log p(x) \approx \log \text{Volume} - \log N + \log \left( \sum_{i=1}^{N} \exp(\log p(x|z_i) + \log p(z_i)) \right)
$$

At this point, the numerically stable `LogSumExp`function can be used to compute the log-likelihood in a more stable manner. The `LogSumExp` function is defined as:
$$
\text{LogSumExp}(x_1, x_2, \ldots, x_n) = \log \left( \sum_{i=1}^{n} e^{x_i} \right)
$$1) + \log p

This function is useful for computing the log of a sum of exponentials in a numerically stable way, since it avoids overflow issues that can arise when exponentiating large values. By using the `LogSumExp` function, we can compute the log-likelihood more accurately and avoid numerical instability in our calculations as follows:
$$
\log p(x) \approx \log \text{Volume} - \log N + \text{LogSumExp}_{i=1}^{N} (\log p(x|z_i) + \log p(z_i))
$$

```python
from torch.distributions import Normal

def estimate_log_px_uniform_rd(vae_model, x, LATENT_DIM, num_samples=1000, bounds=(-5, 5)):
    """
    Estimates log p(x) by generating random uniform samples in R^d within a hypercube,
    using torch.distributions for density evaluation.

    Args:
        vae_model: The trained VAE model.
        x: Input image tensor of shape (1, input_dim).
        LATENT_DIM (int): Dimension 'd' of the latent space.
        num_samples (int): Number of random z samples to generate.
        bounds (tuple): Limits (min, max) for sampling in R^d.

    Returns:
        float: Estimated log p(x).
    """
    device = next(vae_model.parameters()).device
    a, b = bounds

    # 1. Define standard Normal Prior N(0, I) using torch.distributions
    prior = Normal(
        loc=torch.zeros(LATENT_DIM, device=device),
        scale=torch.ones(LATENT_DIM, device=device)
    )

    # 2. Draw 'num_samples' random z vectors uniformly in R^d within range [a, b]
    # z shape: (num_samples, LATENT_DIM)
    z = (b - a) * torch.rand(num_samples, LATENT_DIM, device=device) + a

    # Expand input image 'x' to process all samples in parallel
    x_expanded = x.repeat(num_samples, 1)

    with torch.no_grad():
        # 3. Compute log p(z) using the distribution's log_prob method
        # prior.log_prob(z) returns shape (num_samples, LATENT_DIM), so we sum over dim=1
        log_p_z = prior.log_prob(z).sum(dim=1)

        # 4. Compute log p(x|z) by decoding z
        log_p_xz = vae_model.decoder.get_distribution(z).log_prob(x_expanded).sum(dim=1)

        # 5. Multiply p(x|z) * p(z) -> Addition in log scale
        log_integrand = log_p_xz + log_p_z

        # 6. Apply LogSumExp trick and scale by the volume of the hypercube
        log_volume = LATENT_DIM * torch.log(torch.tensor(b - a, device=device, dtype=torch.float32))

        log_p_x = (
            log_volume
            - torch.log(torch.tensor(num_samples, device=device, dtype=torch.float32))
            + torch.logsumexp(log_integrand, dim=0)
        )

    return log_p_x.item()
```

### 2. Sampling $z\sim p_{\text{prior}}$

Sampling uniformly over a bounded hypercube has a big disadvantage: a lot of the samples drawn from the hypercube will have very low probability under the prior distribution $p(z)$, so all of those samples will contribute very little to the estimate of the marginal likelihood $p(x)$. This can lead to a high variance in the estimator and poor convergence.

Instead of sampling uniformly over a bounded hypercube, a more natural approach is to draw samples directly from the prior distribution $p(z)$, which is defined as a standard Gaussian distribution over the latent space:
$$z \sim \mathcal{N}(0, \mathbf{I})$$

Under this setup, the marginal likelihood is given by:
$$p(x) = \int_{\mathbb{R}^d} p(x|z) p(z) \, dz = \mathbb{E}_{z \sim p(z)}[p(x|z)]$$

Therefore, the marginal likelihood can be approximated using a Monte Carlo estimator:
$$p(x) \approx \frac{1}{N} \sum_{i=1}^{N} p(x|z_i)$$
where $z_i$ are samples drawn from the prior distribution $p(z)$.

The log-likelihood, using the `LogSumExp` function, can be computed as:
$$
\log p(x) \approx -\log N + \text{LogSumExp}_{i=1}^{N} \left( \log p(x|z_i) \right)

```python
def estimate_log_px_prior_sampling(vae_model, x, LATENT_DIM, num_samples=1000):
    """
    Estimates log p(x) by sampling directly from the Gaussian prior p(z) ~ N(0, I),
    using torch.distributions for density evaluation.

    Args:
        vae_model: The trained VAE model.
        x: Input image tensor of shape (1, input_dim).
        LATENT_DIM (int): Dimension 'd' of the latent space.
        num_samples (int): Number of random z samples to generate.

    Returns:
        float: Estimated log p(x).
    """
    device = next(vae_model.parameters()).device

    # 1. Define standard Normal Prior N(0, I)
    prior = Normal(
        loc=torch.zeros(LATENT_DIM, device=device),
        scale=torch.ones(LATENT_DIM, device=device)
    )

    with torch.no_grad():
        # 2. Sample z directly from the Prior: z ~ N(0, I)
        # z shape: (num_samples, LATENT_DIM)
        z = prior.sample((num_samples,))

        # Expand input image 'x' to process all samples in parallel
        x_expanded = x.repeat(num_samples, 1)

        # 3. Compute log p(x|z) by decoding z
        log_p_xz = vae_model.decoder.get_distribution(z).log_prob(x_expanded).sum(dim=1)

        # 4. Apply LogSumExp trick: LogSumExp(log p(x|z)) - log(N)
        log_n = torch.log(torch.tensor(num_samples, device=device, dtype=torch.float32))

        log_p_x = torch.logsumexp(log_p_xz, dim=0) - log_n

    return log_p_x.item()
```

### 3. Importance Sampling

This has however still a big disadvantage: the samples drawn from the prior distribution $p(z)$ may not be representative of the posterior distribution $p(z|x)$, which is the distribution we are actually interested in. This can lead to a high variance in the estimator and poor convergence.

In order to address this issue, we can use importance sampling. Importance sampling allows us to draw samples from a proposal distribution $q(z|x)$ that is closer to the posterior distribution $p(z|x)$. The marginal likelihood can then be estimated as:
$$
p(x) = \int_{\mathbb{R}^d} p(x|z) p(z) \, dz = \int_{\mathbb{R}^d} p(x|z) \frac{p(z)}{q(z|x)} q(z|x) \, dz = \mathbb{E}_{z \sim q(z|x)}\left[ p(x|z) \frac{p(z)}{q(z|x)} \right]
$$

Using the Monte Carlo estimator, we can approximate the marginal likelihood as:
$$
p(x) \approx \frac{1}{N} \sum_{i=1}^{N} p(x|z_i) \frac{p(z_i)}{q(z_i|x)}
$$

Therefore, the log-likelihood can be computed as:
$$
\log p(x) \approx -\log N + \text{LogSumExp}_{i=1}^{N} \left( \log p(x|z_i) + \log p(z_i) - \log q(z_i|x) \right)
$$

```python
def estimate_log_px_importance_sampling(vae_model, x, LATENT_DIM, num_samples=1000):
    """
    Estimates log p(x) using Importance Sampling by drawing samples from
    the proposal distribution q(z|x) (the encoder), and evaluating densities
    with torch.distributions.

    Args:
        vae_model: The trained VAE model.
        x: Input image tensor of shape (1, input_dim).
        LATENT_DIM (int): Dimension 'd' of the latent space.
        num_samples (int): Number of random z samples to generate.

    Returns:
        float: Estimated log p(x) (IWAE / Importance Sampling estimate).
    """
    x = x.view(1, -1)  # Ensure x is of shape (1, input_dim)
    device = next(vae_model.parameters()).device

    # 1. Define standard Normal Prior N(0, I)
    prior = Normal(
        loc=torch.zeros(LATENT_DIM, device=device),
        scale=torch.ones(LATENT_DIM, device=device)
    )

    with torch.no_grad():
        # 2. Get the proposal distribution q(z|x) from the Encoder
        # Assumes vae_model.encoder.get_distribution(x) returns a Normal(mu, std) distribution
        q_zx = vae_model.encoder.get_distribution(x)

        # 3. Sample 'num_samples' z vectors from q(z|x)
        # z shape: (num_samples, LATENT_DIM)
        z = q_zx.sample((num_samples,)).squeeze(1)

        # Expand input image 'x' to process all samples in parallel
        x_expanded = x.repeat(num_samples, 1)

        # 4. Compute log p(x|z) using the Decoder
        log_p_xz = vae_model.decoder.get_distribution(z).log_prob(x_expanded).sum(dim=1)

        # 5. Compute log p(z) under the Prior N(0, I)
        log_p_z = prior.log_prob(z).sum(dim=1)

        # 6. Compute log q(z|x) under the Encoder's proposal
        log_q_zx = q_zx.log_prob(z).sum(dim=1)

        # 7. Compute importance weights: log w_i = log p(x|z_i) + log p(z_i) - log q(z_i|x)
        log_weights = log_p_xz + log_p_z - log_q_zx

        # 8. Apply LogSumExp trick: LogSumExp(log w_i) - log(N)
        log_n = torch.log(torch.tensor(num_samples, device=device, dtype=torch.float32))

        log_p_x = torch.logsumexp(log_weights, dim=0) - log_n

    return log_p_x.item()
```

### 4. ELBO

Although Importance Sampling provides a tighter and unbiased estimate of the marginal log-likelihood $\log p(x)$ as the number of samples $N$ increases, evaluating it during training with a large $N$ can still be computationally expensive. Therefore, we can use the Evidence Lower Bound (ELBO) as a surrogate objective function to optimize the model parameters. The ELBO is defined as:
$$
\text{ELBO} = \mathbb{E}_{z \sim q(z|x)}[\log p(x|z)] - \mathcal{D}_{\text{KL}}(q(z|x) \| p(z))
$$

```python
def estimate_elbo(vae_model, x):
    """
    Estimates the Evidence Lower Bound (ELBO) for a given input x using
    the analytical KL divergence and the reconstruction loss.

    Args:
        vae_model: The trained VAE model.
        x: Input image tensor of shape (1, input_dim).

    Returns:
        float: Estimated ELBO value for the input x.
    """
    device = next(vae_model.parameters()).device

    with torch.no_grad():
        # 1. Pass input through the encoder to get latent distribution parameters
        mu_latent, log_var_latent = vae_model.encoder(x)

        # 2. Sample z using the reparameterization trick (N = 1 sample)
        std_latent = torch.exp(0.5 * log_var_latent)
        eps = torch.randn_like(std_latent)
        z = mu_latent + eps * std_latent

        # 3. Decode z to get the reconstructed mean (mu_reconstruction)
        mu_reconstruction = vae_model.decoder(z)

        # 4. Compute loss using loss_gaussian_vae
        # Note: loss_gaussian_vae returns (total_loss, recon_loss, kl_loss)
        # where total_loss = recon_loss + kl_loss (i.e., -ELBO)
        total_loss, recon_loss, kl_loss = loss_gaussian_vae(
            x=x,
            mu_latent=mu_latent,
            log_var_latent=log_var_latent,
            mu_reconstruction=mu_reconstruction,
            std_reconstruction=0.5
        )

        # Constant C = -D/2 * log(2 * pi * sigma^2), where D is the dimensionality of the input and sigma^2 is the variance of decoder
        D = x.numel()  # Total number of elements in the input tensor
        sigma_squared = 0.5 ** 2  # Variance of the decoder
        C = -0.5 * D * torch.log(torch.tensor(2 * np.pi * sigma_squared, device=device))

        elbo = (-recon_loss+C) - kl_loss

    return elbo.item()
```

### Testing the estimators

Once all the estimators have been implemented, we can test them on a single image in order to compare their performance.

```python
def benchmark_log_px_estimators(vae_model, x_single, latent_dim,
                                sample_sizes=[10, 100, 1000, 10000, 100000],
                                num_trials=10, bounds=(-5, 5)):
    """
    Benchmarks Uniform, Prior, and Importance Sampling log p(x) estimators
    against the analytical ELBO across different sample sizes with visual progress bars.

    Parameters:
    - vae_model: The trained VAE model.
    - x_single: A single input image tensor of shape (1, input_dim).
    - latent_dim: Dimensionality of the latent space.
    - sample_sizes: List of sample sizes (N) to benchmark.
    - num_trials: Number of trials to average results for each method and sample size.
    - bounds: Tuple specifying the bounds for uniform sampling in R^d.

    Returns:
    - results: Dictionary containing mean, std, time, and raw results for each method and sample size.
    - elbo_benchmark: The analytical ELBO value for the input x_single.
    - elbo_benchmark_time: The time taken to compute the ELBO.
    - sample_sizes: The list of sample sizes used in the benchmark.
    """
    vae_model.eval()
    device = next(vae_model.parameters()).device
    x_single = x_single.to(device).view(1, -1)

    # Static ELBO benchmark
    with torch.no_grad():
        start_time = time.time()
        elbo_benchmark = estimate_elbo(vae_model, x_single)
        end_time = time.time()
        elbo_benchmark_time = end_time - start_time

    methods = ['Uniform', 'Prior', 'Importance Sampling']
    results = {m: {N: {'mean': 0.0, 'std': 0.0, 'time': 0.0, 'raw': []} for N in sample_sizes} for m in methods}

    print("Starting Benchmark across trials and sample sizes...")
    print("=" * 65)

    # Outer Progress Bar: Sample Sizes
    pbar_outer = tqdm(sample_sizes, desc="Overall Progress (N)", position=0)

    for N in pbar_outer:
        # Inner Progress Bar: Methods & Trials
        total_steps = len(methods) * num_trials
        pbar_inner = tqdm(total=total_steps, desc=f"Evaluating N = {N:,}", position=1)

        for method in methods:
            trials = []
            start_time = time.time()

            for _ in range(num_trials):
                with torch.no_grad():
                    if method == 'Uniform':
                        val = estimate_log_px_uniform_rd(vae_model, x_single, latent_dim, num_samples=N, bounds=bounds)
                    elif method == 'Prior':
                        val = estimate_log_px_prior_sampling(vae_model, x_single, latent_dim, num_samples=N)
                    elif method == 'Importance Sampling':
                        val = estimate_log_px_importance_sampling(vae_model, x_single, latent_dim, num_samples=N)

                trials.append(val)

                # Update inner progress bar description dynamically
                pbar_inner.set_postfix({
                    "Method": method,
                    "Current Mean": f"{np.mean(trials):.2f}"
                })
                pbar_inner.update(1)

            elapsed = (time.time() - start_time) / num_trials

            results[method][N]['mean'] = np.mean(trials)
            results[method][N]['std'] = np.std(trials)
            results[method][N]['time'] = elapsed
            results[method][N]['raw'] = trials

        pbar_inner.close()

    pbar_outer.close()

    # Print formatted Summary Table for the largest N
    max_N = sample_sizes[-1]
    print("\n" + "=" * 65)
    print(f"BENCHMARK SUMMARY (N = {max_N:,}, {num_trials} Trials):")
    print("=" * 65)
    print(f"{'Method':<22} | {'Mean log p(x)':<15} | {'Std Dev':<10} | {'Avg Time (s)':<12}")
    print("-" * 65)
    print(f"{'ELBO (Lower Bound)':<22} | {elbo_benchmark:<15.4f} | {'N/A':<10} | {elbo_benchmark_time:<12.4f}")
    for m in methods:
        m_mean = results[m][max_N]['mean']
        m_std = results[m][max_N]['std']
        m_time = results[m][max_N]['time']
        print(f"{m:<22} | {m_mean:<15.4f} | {m_std:<10.4f} | {m_time:<12.4f}")
    print("=" * 65)

    return results, elbo_benchmark, elbo_benchmark_time, sample_sizes

# Execute Benchmark
sample_sizes = [int(N) for N in [1e1, 1e2, 1e3, 1e4, 1e5]]
x_single = next(iter(gaussian_testloader))[0][0].view(1, -1)
results, elbo_val, elbo_time, sample_sizes = benchmark_log_px_estimators(
    gaussian_vae, x_single, LATENT_DIM, sample_sizes=sample_sizes, num_trials=100
)
```

```python
def plot_estimator_comparison(results, elbo_val, sample_sizes):
    fig, axes = plt.subplots(1, 3, figsize=(22, 6))
    colors = {'Uniform': '#d62728', 'Prior': '#ff7f0e', 'Importance Sampling': '#1f77b4'}

    # -------------------------------------------------------------------------
    # Plot 1: Convergence vs. Sample Size (N)
    # -------------------------------------------------------------------------
    ax1 = axes[0]
    ax1.axhline(y=elbo_val, color='black', linestyle='--', linewidth=2, label='ELBO (Lower Bound)')

    for method, color in colors.items():
        means = [results[method][N]['mean'] for N in sample_sizes]
        stds = [results[method][N]['std'] for N in sample_sizes]

        ax1.plot(sample_sizes, means, marker='o', color=color, linewidth=2, label=method)
        ax1.fill_between(
            sample_sizes,
            np.array(means) - np.array(stds),
            np.array(means) + np.array(stds),
            color=color, alpha=0.15
        )

    ax1.set_xscale('log')
    ax1.set_xlabel('Number of Monte Carlo Samples ($N$)', fontsize=12)
    ax1.set_ylabel(r'Estimated $\log p(x)$', fontsize=12)
    ax1.set_title('Estimator Convergence vs. Sample Size', fontsize=14, fontweight='bold')
    ax1.grid(True, which="both", linestyle="--", alpha=0.5)
    ax1.legend(fontsize=10)

    # -------------------------------------------------------------------------
    # Plot 2: Estimator Variance (Standard Deviation) Decay
    # -------------------------------------------------------------------------
    ax2 = axes[1]

    for method, color in colors.items():
        stds = [results[method][N]['std'] for N in sample_sizes]
        ax2.plot(sample_sizes, stds, marker='s', color=color, linewidth=2, label=method)

    ax2.set_xscale('log')
    ax2.set_yscale('log')
    ax2.set_xlabel('Number of Monte Carlo Samples ($N$)', fontsize=12)
    ax2.set_ylabel(r'Standard Deviation $\sigma$ (Log Scale)', fontsize=12)
    ax2.set_title('Estimator Variance Reduction Rate', fontsize=14, fontweight='bold')
    ax2.grid(True, which="both", linestyle="--", alpha=0.5)
    ax2.legend(fontsize=10)

    # -------------------------------------------------------------------------
    # Plot 3: Time vs. Sample Size (N)
    # -------------------------------------------------------------------------
    ax3 = axes[2]

    if elbo_time is not None:
        ax3.axhline(y=elbo_time, color='black', linestyle='--', linewidth=2, label='ELBO Time')

    for method, color in colors.items():
        times = [results[method][N]['time'] for N in sample_sizes]
        ax3.plot(sample_sizes, times, marker='^', color=color, linewidth=2, label=method)

    ax3.set_xscale('log')
    ax3.set_yscale('log')
    ax3.set_xlabel('Number of Monte Carlo Samples ($N$)', fontsize=12)
    ax3.set_ylabel('Average Execution Time (s)', fontsize=12)
    ax3.set_title('Computation Time vs. Sample Size', fontsize=14, fontweight='bold')
    ax3.grid(True, which="both", linestyle="--", alpha=0.5)
    ax3.legend(fontsize=10)

    plt.tight_layout()

    plot_path = os.path.join("results", f"Estimator_Comparison.png")
    plt.savefig(plot_path)
    plt.show()


plot_estimator_comparison(results, elbo_val, sample_sizes)
```

Looking at the benchmark results, the ELBO is clearly the best baseline: it provides a very tight approximation of $\log p(x)$, has literally zero variance (since it's deterministic), and is way faster to compute than any sampling approach.


As for the Monte Carlo methods, Importance Sampling dominates Uniform and Prior sampling across the board. It converges much faster to the actual log-likelihood with drastically reduced variance, while taking virtually the same runtime as the other sampling methods.
